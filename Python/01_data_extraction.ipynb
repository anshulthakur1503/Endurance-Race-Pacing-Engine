```python
import mysql.connector
import pandas as pd

# 1. Establish the connection
try:
    connection = mysql.connector.connect(
        host="localhost",
        user="root",          # Replace with your MySQL username if different
        password="anything", # Replace with your actual MySQL password
        database="race_pacing_engine"
    )
    
    if connection.is_connected():
        print("✅ Successfully connected to MySQL database!")

    # 2. Write the SQL query to extract data
    # We are pulling the demographics and finish times for Race 1 (Half Marathon)
    query = """
    SELECT 
        vw.FirstName, 
        vw.LastName, 
        vw.Gender, 
        vw.AgeOnRaceDay, 
        vw.Category,
        sl.Chip_Timestamp AS FinishTime
    FROM vw_runner_demographics vw
    JOIN SplitLogs sl ON vw.RegID = sl.RegID
    WHERE vw.RaceID = 1 AND sl.Kilometer_Mark = 21.10;
    """

    # 3. Read the data into a Pandas DataFrame
    df_finishers = pd.read_sql(query, connection)
    
    print("\nData Extraction Successful! Here are the first 5 rows:")
    display(df_finishers.head())

except mysql.connector.Error as e:
    print(f"❌ Error connecting to MySQL: {e}")

finally:
    # 4. Always close the connection when done
    if 'connection' in locals() and connection.is_connected():
        connection.close()
        print("\n🔒 MySQL connection is closed.")
```
